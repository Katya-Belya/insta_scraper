# OCR Smoke Test

This notebook verifies that the local OCR environment is working.
This notebook is intended to verify the OCR environment rather than maximize extraction accuracy.

It confirms that:

- the sample image can be loaded,
- basic preprocessing can be applied,
- Python can communicate with Tesseract,
- and OCR returns text successfully.

Accurate field extraction is tested separately in
`02_region_ocr_test.ipynb`.

```python
from pathlib import Path

from PIL import Image
import pytesseract


image_path = Path("../data/sample/cherry_blossom_market.jpeg")

image = Image.open(image_path)
text = pytesseract.image_to_string(image)

print("----- OCR OUTPUT -----")
print(text)
```

## Smoke Test Result

The OCR pipeline successfully:

- loaded the sample image,
- executed Tesseract,
- and returned readable text without errors.

While the output still contains recognition errors, the primary event information (date and time) was successfully extracted. More accurate extraction is explored in `02_region_ocr_test.ipynb`.
